```python
import numpy as np
import matplotlib.pyplot as plt
```

2 non-vehicle currencies + vehicle case

```python
def f1(v):
    return np.sqrt(2 * (1 + 2 * v) / (1 + v))

def f2(v):
    return 2 * (np.sqrt(1 + v) - 1) / np.sqrt(v)

def f3(v):
    return np.sqrt(2 * (1 + 2 * v) / (2 + v))

# range of v values
v_values = np.linspace(0.01, 10, 500) # start from a small positive value to avoid issues with v in denominator

# calculate y values for each function for each v
y1_values = f1(v_values)
y2_values = f2(v_values)
y3_values = f3(v_values)

plt.figure(figsize=(8, 5))

# shade the regions with different textured patterns and colors
plt.fill_between(v_values, y1_values, facecolor='lightblue', edgecolor='darkblue', alpha=0.3, label=r'$c^* < c^\text{SQ}$')
plt.fill_between(v_values, y2_values, facecolor='none', edgecolor='red', hatch='.', alpha=0.5, label=r'$c^\text{BP} < c^\text{SQ}$')
plt.fill_between(v_values, y3_values, facecolor='none', edgecolor='green', hatch='/', alpha=0.5, label=r'$c^\text{SYM} < c^\text{SQ}$')

plt.xlabel('v')
plt.ylabel('s')
plt.xlim(0,10)
plt.legend(framealpha=0.95)
plt.ylim(bottom=0)
plt.savefig('output/param_regions.png', dpi=300, bbox_inches="tight")
plt.show()
```

N non-vehicle currencies + vehicle case

```python
N = 5

def f1(v):
    return np.sqrt(2 * (1 + N * v) / (1 + (N-1)* v))

def f2(v):
    return 2 * (np.sqrt(1 + (N-1)*v) - 1) /((N-1)* np.sqrt(v))

def f3(v):
    return np.sqrt(2 * (1 + N * v) / (2 + (N-1)*v))

# range of v values
v_values = np.linspace(0.01, 3, 500) # start from a small positive value to avoid issues with v in denominator

# calculate threshold for each function for each v
y1_values = f1(v_values)
y2_values = f2(v_values)
y3_values = f3(v_values)

# Plot the functions
plt.figure(figsize=(8, 5))

# shade the regions with different textured patterns and colors
plt.fill_between(v_values, y1_values, facecolor='lightblue', edgecolor='darkblue', alpha=0.3, label=r'$c^* < c^\text{SQ}$')
plt.fill_between(v_values, y2_values, facecolor='none', edgecolor='red', hatch='.', alpha=0.5, label=r'$c^\text{BP} < c^\text{SQ}$')
plt.fill_between(v_values, y3_values, facecolor='none', edgecolor='green', hatch='/', alpha=0.5, label=r'$c^\text{SYM} < c^\text{SQ}$')
plt.axhline(y=np.sqrt(2*N/(N-1)), color='darkblue', linestyle='--', linewidth=0.7, label=r"limit for $c^*$ threshold")
plt.axhline(y=2/np.sqrt(N-1), color='red', linestyle='--', linewidth=0.7, label=r"limit for $c^\text{BP}$ threshold")

plt.xlabel('v')
plt.ylabel('s')
plt.xlim(0,3)
plt.legend(framealpha=0.95)
plt.ylim(bottom=0)
plt.savefig('output/param_regions_N.png', dpi=300, bbox_inches="tight")
plt.show()
```

threshold limits as v goes to infinity for variety of N

```python
def threshold_star(N):
  return np.sqrt(2*N/(N-1))
def threshold_BP(N):
  return 2/np.sqrt(N-1)

# range of v values
N_values = np.linspace(2, 10, 500) # start from 2

# calculate threshold limit for each function
y1_values = threshold_star(N_values)
y2_values = threshold_BP(N_values)

plt.figure(figsize=(6, 5))

# shade the regions with different textured patterns and colors
plt.plot(N_values, y1_values, color='darkblue', label=r"$c^*$ and $c^\text{SYM}$")
plt.plot(N_values, y2_values, color='red', label=r"$c^\text{BP}$")

plt.xlabel('N')
plt.ylabel(r'Limit of threshold for $s$ as $v \rightarrow \infty$')
plt.xlim(2,10)
plt.legend(loc='lower right') 
plt.ylim(bottom=0)
plt.savefig('output/thresholds.png', dpi=300, bbox_inches="tight")
plt.show()
```
